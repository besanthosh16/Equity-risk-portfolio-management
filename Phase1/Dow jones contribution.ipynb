```python
import pandas as pd
import yfinance as yf

# 1. Dow Jones 10 Tickers
dow_30_tickers = ['GS', 'CAT','UNH','TRV','MSFT','AMGN','V','JPM','AAPL','HD']
print("Downloading daily close prices for Dow 10...")

# 2. Download data (last 10 years)
df = yf.download(
    tickers=dow_30_tickers,
    start="2016-07-29",
    end="2026-07-29",
    interval='1d',
    auto_adjust=True
)

# 3. Extract ONLY the 'Close' prices matrix
close_matrix = df['Close'].copy()

# 4. Clean formatting (Date as column, remove extra index names)
close_matrix = close_matrix.reset_index()
close_matrix.columns.name = None  # Removes multi-level column label headers

# 5. Save to CSV
output_filename = "market risk model(10).csv"
close_matrix.to_csv(output_filename, index=False)

print(f"Data saved successfully to {output_filename}!")

# Preview top rows
print("\nPreview of Wide Format (Date + Tickers across top):")
print(close_matrix.head())
```
