```python
import pandas as pd
import yfinance as yf
from datetime import datetime, timedelta

def fetch_us_stock_data(tickers, period="10y", interval="1d", save_to_csv=True):
    
    print(f"Fetching data for: {', '.join(tickers)}...")
    
    # Download data using yfinance
    df = yf.download(
        tickers=tickers,
        period=period,
        interval=interval,
        group_by='ticker',
        auto_adjust=True,  # Adjusts for splits and dividends automatically
        threads=True       # Speeds up multi-ticker downloads
    )
    
    # Restructure data into a flat, clean Pandas DataFrame
    data_list = []
    
    # Handle both single ticker and multiple ticker responses
    if len(tickers) == 1:
        ticker = tickers[0]
        df_single = df.copy()
        df_single['Ticker'] = ticker
        df_single = df_single.reset_index()
        data_list.append(df_single)
    else:
        for ticker in tickers:
            if ticker in df:
                df_single = df[ticker].dropna().copy()
                df_single['Ticker'] = ticker
                df_single = df_single.reset_index()
                data_list.append(df_single)
    
    # Combine into a single clean DataFrame
    clean_df = pd.concat(data_list, ignore_index=True)
    
    # Reorder columns for readability
    cols = ['Date', 'Ticker', 'Open', 'High', 'Low', 'Close', 'Volume']
    clean_df = clean_df[[c for c in cols if c in clean_df.columns]]
    
    if save_to_csv:
        filename = f"us_stocks_data_{datetime.now().strftime('%Y%m%d')}.csv"
        clean_df.to_csv(filename, index=False)
        print(f"Data saved successfully to {filename}")
        
    return clean_df

# --- Example Usage ---
if __name__ == "__main__":
    # Define US stock tickers (Dow jones / Tech leaders example)
    stock_list = ['GS', 'CAT','UNH','TRV','MSFT','AMGN','V','JPM','AAPL','HD']
    
    # Download 10 year of daily data
    stock_data = fetch_us_stock_data(
        tickers=stock_list,
        period="10y",
        interval="1d",
        save_to_csv=True
    )
    
    # Display the first few rows
    print("\nPreview of collected data:")
    print(stock_data.head())
```
