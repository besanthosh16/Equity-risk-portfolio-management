```python
import pandas as pd
import yfinance as yf

# Download historical data for the Dow Jones Index
dji_data = yf.download(
    tickers='^DJI',
    period='10y',
    interval='1d',
    auto_adjust=True
)

# Clean up index formatting
dji_data = dji_data.reset_index()
dji_data['Ticker'] = 'Dow Jones'

print(dji_data.head())
dji_data.to_csv('dow_jones_index_data.csv', index=False)
```
