```python
import pandas as pd
import yfinance as yf
from datetime import datetime, timedelta

def fetch_us_stock_data(tickers, period="10y", interval="1d", save_to_csv=True):
    
    print(f"Fetching data for: {', '.join(tickers)}...")
    
    # Download data using yfinance
    df = yf.download(
        tickers=tickers,
        period=period,
        interval=interval,
        group_by='ticker',
        auto_adjust=True,  # Adjusts for splits and dividends automatically
        threads=True       # Speeds up multi-ticker downloads
    )
    
    # Restructure data into a flat, clean Pandas DataFrame
    data_list = []
    
    # Handle both single ticker and multiple ticker responses
    if len(tickers) == 1:
        ticker = tickers[0]
        df_single = df.copy()
        df_single['Ticker'] = ticker
        df_single = df_single.reset_index()
        data_list.append(df_single)
    else:
        for ticker in tickers:
            if ticker in df:
                df_single = df[ticker].dropna().copy()
                df_single['Ticker'] = ticker
                df_single = df_single.reset_index()
                data_list.append(df_single)
    
    # Combine into a single clean DataFrame
    clean_df = pd.concat(data_list, ignore_index=True)
    
    # Reorder columns for readability
    cols = ['Date', 'Ticker', 'Close']
    clean_df = clean_df[[c for c in cols if c in clean_df.columns]]
    
    if save_to_csv:
        filename = f"us_stocks_data_{datetime.now().strftime('%Y%m%d')}.csv"
        clean_df.to_csv(filename, index=False)
        print(f"Data saved successfully to {filename}")
        
    return clean_df

# --- Example Usage ---
if __name__ == "__main__":
    # Define US stock tickers (Dow jones / Tech leaders example)
    stock_list = ['GS', 'CAT','UNH','TRV','MSFT','AMGN','V','JPM','AAPL','HD']
    
    # Download 10 year of daily data
    stock_data = fetch_us_stock_data(
        tickers=stock_list,
        period="10y",
        interval="1d",
        save_to_csv=True
    )
    
    # Display the first few rows
    print("\nPreview of collected data:")
    print(stock_data.head())
```

```python
import pandas as pd
import numpy as np

# ==========================================
# STEP 1: Load Data & Sort Chronologically
# ==========================================
# Replace with your actual filename if different
df = pd.read_csv("us_stocks_data_20260729.csv") 
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values(['Ticker', 'Date']).reset_index(drop=True)

# ==========================================
# STEP 2: Create Clean Price Matrix
# ==========================================
# Pivot long DataFrame to a wide Price Matrix
price_matrix = df.pivot(index='Date', columns='Ticker', values='Close')

# Forward-fill gaps, backward-fill initial missing values
price_matrix = price_matrix.ffill().bfill()

# Remove stale flat prices (zero returns)
price_matrix = price_matrix.mask(price_matrix.pct_change() == 0).ffill()

# ==========================================
# STEP 3: Log Returns & Winsorization
# ==========================================
# Calculate Log Returns
log_returns = np.log(price_matrix / price_matrix.shift(1)).dropna()

# Pure Pandas Winsorization function (caps bottom/top 0.5% outliers)
def winsorize_pandas(series, lower_quantile=0.005, upper_quantile=0.005):
    lower = series.quantile(lower_quantile)
    upper = series.quantile(1 - upper_quantile)
    return series.clip(lower=lower, upper=upper)

clean_return_matrix = log_returns.apply(winsorize_pandas, axis=0)

# ==========================================
# STEP 4: Build Portfolio Returns
# ==========================================
# Equal-weighted portfolio (10% per stock)
num_stocks = clean_return_matrix.shape[1]
weights = np.array([1.0 / num_stocks] * num_stocks)

portfolio_returns = clean_return_matrix.dot(weights)
portfolio_returns.name = "Portfolio_Log_Return"

# ==========================================
# STEP 5: Rolling Volatility Series
# ==========================================
# 21-day annualized rolling volatility
rolling_window = 21
annual_factor = np.sqrt(252)

rolling_volatility = portfolio_returns.rolling(window=rolling_window).std() * annual_factor
rolling_volatility.name = "Rolling_21D_Vol_Annualized"

# ==========================================
# STEP 6: Save Output Files
# ==========================================
price_matrix.to_csv("clean_price_matrix.csv")
clean_return_matrix.to_csv("clean_return_matrix.csv")

portfolio_summary = pd.concat([portfolio_returns, rolling_volatility], axis=1)
portfolio_summary.to_csv("portfolio_performance.csv")

print("All tasks completed successfully!")
print("\nPreview of Portfolio Returns & Volatility:")
print(portfolio_summary.tail())
```
